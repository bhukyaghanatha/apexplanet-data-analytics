```python
query = """
SELECT category, COUNT(*) AS order_count
FROM ecommerce_sales
GROUP BY category
ORDER BY order_count DESC;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    SUM(total_amount) AS total_sales,
    AVG(total_amount) AS average_order_value,
    MIN(total_amount) AS minimum_order_value,
    MAX(total_amount) AS maximum_order_value
FROM ecommerce_sales;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    category,
    COUNT(*) AS order_count,
    AVG(total_amount) AS average_sales
FROM ecommerce_sales
GROUP BY category
HAVING COUNT(*) > 500
ORDER BY order_count DESC;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
CREATE TABLE IF NOT EXISTS category_summary AS
SELECT
    category,
    COUNT(*) AS order_count,
    SUM(total_amount) AS total_sales
FROM ecommerce_sales
GROUP BY category;
"""

conn.execute(query)
conn.commit()

print("category_summary table created!")
```

```python
pd.read_sql("SELECT * FROM category_summary", conn)
```

```python
query = """
SELECT
    e.order_id,
    e.category,
    e.total_amount,
    c.order_count,
    c.total_sales
FROM ecommerce_sales AS e
INNER JOIN category_summary AS c
    ON e.category = c.category
LIMIT 10;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    e.order_id,
    e.category,
    e.total_amount,
    c.order_count,
    c.total_sales
FROM ecommerce_sales AS e
LEFT JOIN category_summary AS c
    ON e.category = c.category
LIMIT 10;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    order_id,
    category,
    total_amount
FROM ecommerce_sales
WHERE total_amount > (
    SELECT AVG(total_amount)
    FROM ecommerce_sales
)
ORDER BY total_amount DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
WITH category_sales AS (
    SELECT
        category,
        SUM(total_amount) AS total_sales
    FROM ecommerce_sales
    GROUP BY category
)
SELECT
    category,
    total_sales
FROM category_sales
ORDER BY total_sales DESC;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    order_id,
    category,
    total_amount,
    ROW_NUMBER() OVER (
        PARTITION BY category
        ORDER BY total_amount DESC
    ) AS row_num
FROM ecommerce_sales
LIMIT 20;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    order_id,
    category,
    total_amount,
    RANK() OVER (
        PARTITION BY category
        ORDER BY total_amount DESC
    ) AS rank_num
FROM ecommerce_sales
LIMIT 20;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    order_id,
    order_date,
    total_amount,
    LAG(total_amount) OVER (
        ORDER BY order_date
    ) AS previous_order_amount
FROM ecommerce_sales
LIMIT 20;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    order_id,
    order_date,
    total_amount,
    LEAD(total_amount) OVER (
        ORDER BY order_date
    ) AS next_order_amount
FROM ecommerce_sales
LIMIT 20;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    strftime('%Y-%m', order_date) AS month,
    SUM(total_amount) AS monthly_sales
FROM ecommerce_sales
GROUP BY month
ORDER BY month;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    customer_id,
    SUM(total_amount) AS total_revenue
FROM ecommerce_sales
GROUP BY customer_id
ORDER BY total_revenue DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
WITH customer_orders AS (
    SELECT
        customer_id,
        COUNT(*) AS order_count
    FROM ecommerce_sales
    GROUP BY customer_id
)
SELECT
    COUNT(*) AS total_customers,
    SUM(CASE WHEN order_count > 1 THEN 1 ELSE 0 END) AS returning_customers,
    ROUND(
        100.0 * SUM(CASE WHEN order_count > 1 THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS retention_rate_percent
FROM customer_orders;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    category,
    COUNT(*) AS total_orders,
    SUM(total_amount) AS total_sales,
    AVG(total_amount) AS average_order_value,
    AVG(profit_margin) AS average_profit_margin
FROM ecommerce_sales
GROUP BY category
ORDER BY total_sales DESC;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    order_date,
    total_amount,
    AVG(total_amount) OVER (
        ORDER BY order_date
        ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
    ) AS moving_avg_7_orders,
    SUM(total_amount) OVER (
        ORDER BY order_date
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS cumulative_sales
FROM ecommerce_sales
ORDER BY order_date
LIMIT 20;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT
    order_date,
    total_amount,
    AVG(total_amount) OVER (
        ORDER BY order_date
        ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
    ) AS moving_avg_7_orders,
    SUM(total_amount) OVER (
        ORDER BY order_date
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS cumulative_sales
FROM ecommerce_sales
ORDER BY order_date
LIMIT 20;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
CREATE VIEW IF NOT EXISTS category_sales_view AS
SELECT
    category,
    COUNT(*) AS total_orders,
    SUM(total_amount) AS total_sales,
    AVG(total_amount) AS average_order_value
FROM ecommerce_sales
GROUP BY category;
"""

conn.execute(query)
conn.commit()

print("View created successfully!")
```

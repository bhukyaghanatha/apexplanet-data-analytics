```python
numeric_cols = [
    "price",
    "discount",
    "quantity",
    "delivery_time_days",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "customer_age"
]

Q1 = df_clean[numeric_cols].quantile(0.25)
Q3 = df_clean[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_counts = (
    ((df_clean[numeric_cols] < lower_bound) |
     (df_clean[numeric_cols] > upper_bound))
    .sum()
)

outlier_counts
```

```python
outlier_summary = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower_Bound": lower_bound,
    "Upper_Bound": upper_bound,
    "Outlier_Count": outlier_counts
})

outlier_summary
```

```python
df_clean_capped = df_clean.copy()

for col in numeric_cols:
    df_clean_capped[col] = df_clean_capped[col].clip(
        lower=lower_bound[col],
        upper=upper_bound[col]
    )
```

```python
df_final = df_clean.copy()

continuous_cols = [
    "price",
    "total_amount",
    "shipping_cost",
    "profit_margin"
]

for col in continuous_cols:
    df_final[col] = df_final[col].clip(
        lower=lower_bound[col],
        upper=upper_bound[col]
    )
```

```python
# Make a copy of the original dataset
df_final = df.copy()

# Convert order_date to datetime
df_final["order_date"] = pd.to_datetime(df_final["order_date"])

# Remove duplicate rows
df_final = df_final.drop_duplicates()

# Apply the same IQR capping you already performed
numeric_cols = [
    "price",
    "discount",
    "quantity",
    "delivery_time_days",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "customer_age"
]

for col in numeric_cols:
    Q1 = df_final[col].quantile(0.25)
    Q3 = df_final[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    df_final[col] = df_final[col].clip(lower=lower, upper=upper)
```
